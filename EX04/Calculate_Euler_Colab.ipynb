### Calculo de Euler

```python
import os, sys, math
import matplotlib.pyplot as plt
```

```python
def Calculate_precision(decimal_value):
    return 10 ** (-decimal_value)
```

```python
def Calculate_euler(precision):
  euler_approx = 0.0
  TERM = 1.0
  interation_count = 0

  interation_list = []
  euler_values = []

  print("Interação | Valor de Euler")
  print("----------|----------------")

  while TERM > precision:
    euler_approx += TERM
    interation_count += 1

    interation_list.append(interation_count)
    euler_values.append(euler_approx)
    print(f"{interation_count} | {euler_approx:.10f}")

    TERM /= interation_count

  return euler_approx, interation_count, interation_list, euler_values
```

```python
decimal_value_precision = int(input("Digite a precisão desejada (número de casas decimais): "))
precision_value = Calculate_precision(int(decimal_value_precision))

EULER, interation_nedded, interation_histotry, euler_history = Calculate_euler(precision_value)

print(f"Valor de Euler: {EULER:.10f} Quantidade de iterações: {interation_nedded} Precisão: {precision_value}")
```

```python
plt.figure(figsize=(8, 5))
plt.plot(interation_histotry, euler_history, 'b.-', label='Aproximação de Euler')
plt.axhline(y=math.e, color='r', linestyle='--', label='Valor real de Euler')
plt.xlabel('Número de Iterações')
plt.ylabel('Valor de Euler')
plt.legend()
plt.grid(True)
plt.show()
```
